```julia
#Obtenir toutes les notes

frequence = (65, 69, 74, 78, 83, 87, 93, 98, 104, 110, 117, 123)
notes=[]
samplerate = 44100
note = ["C", "C_diese", "D", "D_diese", "E", "F", "F_diese", "G", "G_diese", "A", "A_diese", "B"]
#Plot le graphe
Δt = 1/samplerate
temps = 0:Δt:1.0

for i in frequence
    for k in note
        signal = sin.(2*π*i*temps)

        wavwrite(Float32.(signal), "/home/djangohulot/Downloads/sonifSismo.jl-main/notebooks/notes/$(k)Hz.wav", Fs=samplerate )
    
        push!(notes,signal)

    end
end
```

```julia
function helper_get_noteArray(noteNameHere::String,notes)
    noteNames = ["C", "C_diese", "D", "D_diese", "E", "F", "F_diese", "G", "G_diese", "A", "A_diese", "B"]
    indexNote=nothing
    for i in eachindex(noteNames)
        if noteNameHere == noteNames[i]
            indexNote=i
        end
    end
    return notes[indexNote]
end

```
