# example of reading one trace

```julia
using Pkg

cd(@__DIR__)
Pkg.activate("../")
using Dates, SonifSismo, CairoMakie
```

```julia
archive = DataArchive("C:/Users/lucie/Desktop/mtFujiContinuous/mtFujiContinuous")
```

```julia

available_stations(archive)
available_channels(archive)
```

```julia
start_time = DateTime(2008, 5, 29, 12-9, 0, 0)  # UTC
end_time   = DateTime(2008, 5, 29, 15-9, 0, 0)     # UTC

# TEST 1 - de 0,5 à 3 Hz

traces = read_window(
    archive,
    start_time,
    end_time;
    stations="FUJ",
    channels=["wE", "wN", "wU"],
    merge=true,
 #    merge_gaps=:zero,       # or :linear
    processed=true,
    bandpass=(0.5, 3),
)

# Graphe
plot_events = read_catalog(      # Lire les événements pour la période spécifiée
    archive;
    start=start_time,
    stop=end_time,
)

wavefig = plot_waveforms(        # Afficher les formes d'ondes des traces
    traces;
    events=plot_events,
    title="FUJ_29 mai 2008, 12h-15h (JST), bande passante 0.5-3 Hz",
)
wavefig

save("C:/Users/lucie/Desktop/Sonification/passe_bande_0.5-3Hz.png", wavefig)

# Sonification
include("C:/Users/lucie/Desktop/sonifSismo.jl-main/examples/sonify_one_trace.jl")
trace_wU = only(t for t in traces if strip(t.sta.cha) == "wU")

audio = sonify_trace(
    trace_wU;
    acceleration=128,
    gain=4,
    saturation=:soft,
)
write_sonification("audio/0,5-3Hz.wav", audio)
```
